```python
def relu(x):
    out = np.zeros((len(x),1))
    for i in range(len(x)):
        out[i] = max(0,x[i])
    return out
```

```python
def metric(x1, y1, x2, y2):
    temp = np.sort(np.unique(np.union1d(x1,x2)))
    v = np.linspace(temp[0],temp[-1],10000)

    c1 = np.interp(v,x1,y1)
    c2 = np.interp(v,x2,y2)
    area = np.sqrt(np.mean((c1 - c2)**2))

    return area
```

```python
def reconstruct(P, Q, K, curve):
    v = np.hstack((0,P,np.linspace(Q,K,len(curve))))
    p = np.hstack((3000,3000,curve))
    recon_curve = np.vstack((v,p))
    return recon_curve

def reconstruct_sup(P, Q, K, curve):
    v = np.hstack((np.linspace(P,Q,len(curve)),K, K + 200))
    p = np.hstack((curve,3000,3000))
    recon_curve = np.vstack((v,p))
    return recon_curve
```

```python
def mean_curve(vol, price):
    v = []
    for i in range(len(vol)):
        v = np.unique(np.union1d(v, vol[i]))
    v = np.sort(v)
    crv = []
    for i in range(len(vol)):
        crv.append(np.interp(v, vol[i], price[i]))
    return (v, np.mean(np.array(crv),axis=0))
```

```python
def step_interpolation(Vi, Pi, Vx):
    # Find the index of the closest Vi value to each Vx value
    idx = np.searchsorted(Vi, Vx, side='right') - 1
    idx[idx == -1] = 0
    # Interpolate using the left side step for each Vx value
    interpolated_Px = Pi[idx]
    
    return interpolated_Px

def metric(v1, p1, v2, p2):
    #com_v = np.sort(np.union1d(v1, v2))
    com_v = np.linspace(0, v1[-1],10000)
    c1 = step_interpolation(v1, p1, com_v)
    c2 = step_interpolation(v2, p2, com_v)
    
    return(np.mean(np.abs(c1 - c2)))
```

```python
def reconstruct_sup(P, Q, K, curve):
    v = np.hstack((np.linspace(P,Q,len(curve)),K, K + 200))
    p = np.hstack((curve,3000,3000))
    recon_curve = np.vstack((v,p))
    return recon_curve
```

```python
def right_step_interpolation(x, y, x_new):
    # Use numpy's searchsorted to find the indices where elements of x_new would be inserted into x
    indices = np.searchsorted(x, x_new, side='left')
    # Clip indices to stay within valid range
    indices = np.clip(indices, 0, len(x) - 1)
    # Use the indices to select corresponding y values
    y_new = y[indices]
    
    return y_new

def metric(v1, p1, v2, p2):
    #com_v = np.sort(np.union1d(v1, v2))
    com_v = np.linspace(0, v1[-1],10000)
    c1 = right_step_interpolation(v1, p1, com_v)
    c2 = right_step_interpolation(v2, p2, com_v)
    
    return(np.mean(np.abs(c1 - c2)))
```

```python
def step_interpolation(Vi, Pi, Vx):
    # Find the index of the closest Vi value to each Vx value
    idx = np.searchsorted(Vi, Vx, side='right') - 1
    idx[idx == -1] = 0
    # Interpolate using the left side step for each Vx value
    interpolated_Px = Pi[idx]
    
    return interpolated_Px

def metric(v1, p1, v2, p2):
    #com_v = np.sort(np.union1d(v1, v2))
    com_v = np.linspace(0, v1[-1],10000)
    c1 = step_interpolation(v1, p1, com_v)
    c2 = step_interpolation(v2, p2, com_v)
    
    return(np.mean(np.abs(c1 - c2)))
```
